```python
import matplotlib.pyplot as plt

restaurants = ["Restaurant A", "Restaurant B", "Restaurant C", "Restaurant D", "Restaurant E"]
orders = [120, 150, 90, 180, 130]

plt.figure(figsize=(8, 5))
plt.bar(restaurants, orders, color="orange")

plt.title("Zomato Restaurant Orders")
plt.xlabel("Restaurants")
plt.ylabel("Number of Orders")

plt.savefig("zomato_orders.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [4000, 5000, 6500, 7000, 6000, 9000, 8000]

plt.figure(figsize=(8, 5))
plt.plot(days, steps, marker="o", color="green")

plt.title("Daily Step Count")
plt.xlabel("Days")
plt.ylabel("Number of Steps")

plt.savefig("daily_steps.svg", bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt

apps = ["Instagram", "YouTube", "WhatsApp", "Spotify"]
hours = [3, 4, 2, 1]

plt.figure(figsize=(7, 7))
plt.pie(hours, labels=apps, autopct="%1.1f%%", startangle=90)

plt.title("Daily App Usage")

plt.savefig("app_usage.pdf", bbox_inches="tight")
plt.show()
```
